## Partie 2 : Correlation

```python
# Question 1:
def cov(x, y):
    total_elements = len(x)
    mean_val_x = mean(x)
    mean_val_y = mean(y)
    sum_product_xy = sum(x[i] * y[i] for i in range(total_elements)) / total_elements
    return sum_product_xy - mean_val_x * mean_val_y
```

```python
covariance_result = cov(num_friends, dayly_minutes)
print(covariance_result)
```

```python
std_dev_x = std(num_friends)
std_dev_y = std(dayly_minutes)
correlation_coeff = covariance_result / (std_dev_x * std_dev_y)
print(correlation_coeff)
```

```python
def quantile(data, p):
    p_index = int(p * len(data))
    return sorted(data)[p_index]
```

```python
# question 1:

def QQ_plot(X, Y, p):
    quantiles_x = []
    quantiles_y = []
    q = p
    while q < 1:
        quantiles_x.append(quantile(X, q))
        quantiles_y.append(quantile(Y, q))
        q += p
    points_qq = list(zip(quantiles_y, quantiles_x))
    g = scatter_plot(points_qq, markersize=15, facecolor='blue')

    val_min = min(min(quantiles_x), min(quantiles_y))
    val_max = max(max(quantiles_x), max(quantiles_y))
    x_var = var('x')
    bissectrice = plot(x_var, (x_var, val_min, val_max), color='green', linestyle='--')

    return g + bissectrice
```
