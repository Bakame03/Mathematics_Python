```python
# pour utiliser matplotlib
import numpy as np # pour les maths
import pandas as pd # pour l'analyse de données
import matplotlib.pyplot as plt # pour les graphiques
```

```python
# 2
iris = pd.read_csv("iris.csv", sep=",", encoding="utf-8")
```

```python
iris.head()
```

```python
iris.corr(numeric_only=True)
```

```python
sepal_length = iris.loc[:, "sepal_length"]
sepal_width = iris.loc[:, "sepal_width"]
lab = iris.loc[:, "species"]

# Affichage des données
for e, c in [('setosa','g'), ('virginica', 'r'), ('versicolor', 'b')]:
    plt.scatter(sepal_length[lab == e], sepal_width[lab == e], color = c, label = e)

plt.legend()
plt.show()
    
```

```python
# Variables 
lab = iris.loc[:, "species"]

# Liste des paires à tracer (en excluant sepal_length vs sepal_width déjà fait)
paires_a_tracer = [
    ("petal_length", "sepal_length"),
    ("petal_width", "sepal_length"),
    ("petal_length", "sepal_width"),
    ("petal_width", "sepal_width"),
    ("petal_length", "petal_width")
]

for axe_x, axe_y in paires_a_tracer:
    # Sélection des données pour les axes
    x_data = iris.loc[:, axe_x]
    y_data = iris.loc[:, axe_y]
    
    # Création de la figure
    plt.figure()
    
    # Affichage des données 
    for e, c in [('setosa','g'), ('virginica', 'r'), ('versicolor', 'b')]:
        plt.scatter(x_data[lab == e], y_data[lab == e], color = c, label = e)
    
    # Habillage du graphique
    plt.xlabel(axe_x)
    plt.ylabel(axe_y)
    plt.title(f"Nuage de points : {axe_y} en fonction de {axe_x}")
    plt.legend()
    plt.show()
```

```python
point_moyen = iris.mean(numeric_only=True)

print("Coordonnées du point moyen :")
print(point_moyen)
```

```python
# Calcul des coordonnées du point moyen 
point_moyen = iris.mean(numeric_only=True)
x_moyen = point_moyen["petal_length"]
y_moyen = point_moyen["petal_width"]

# Préparation des données pour le graphique de la Figure 3
axe_x = "petal_length"
axe_y = "petal_width"
lab = iris.loc[:, "species"]
x_data = iris.loc[:, axe_x]
y_data = iris.loc[:, axe_y]

# Création du graphique
plt.figure()

# Affichage des points du jeu de données par espèce
for e, c in [('setosa','g'), ('virginica', 'r'), ('versicolor', 'b')]:
    plt.scatter(x_data[lab == e], y_data[lab == e], color = c, label = e)

# AJOUT DU POINT MOYEN 
# ici j'utilise un marqueur 'X' noir pour qu'il soit bien visible
plt.scatter(x_moyen, y_moyen, color='black', marker='X', s=100, label='Point Moyen')

# Habillage
plt.xlabel(axe_x)
plt.ylabel(axe_y)
plt.title(f"Figure 3 : {axe_y} en fonction de {axe_x} avec Point Moyen")
plt.legend()
plt.show()

print(f"Le point moyen est projeté aux coordonnées : x={x_moyen:.3f}, y={y_moyen:.3f}")
```

```python
# Nouvelles dimensions 
x_new = 2.0
y_new = 0.5

# Préparation des données pour le graphique de la Figure 3
axe_x = "petal_length"
axe_y = "petal_width"
lab = iris.loc[:, "species"]
x_data = iris.loc[:, axe_x]
y_data = iris.loc[:, axe_y]

# Création du graphique
plt.figure()

# Affichage des points du jeu de données par espèce
for e, c in [('setosa','g'), ('virginica', 'r'), ('versicolor', 'b')]:
    plt.scatter(x_data[lab == e], y_data[lab == e], color = c, label = e)

# AJOUT DU NOUVEL IRIS 
# ici j'utilise le marqueur 'X' pour situer l'iris (2.0, 0.5)
plt.scatter(x_new, y_new, color='black', marker='X', s=100, label='Nouvel Iris')

# Habillage
plt.xlabel(axe_x)
plt.ylabel(axe_y)
plt.title(f"Figure 3 : {axe_y} en fonction de {axe_x} (Nouvel Iris : 2.0, 0.5)")
plt.legend()
plt.show()

print(f"Le nouvel iris est projeté aux coordonnées : x={x_new}, y={y_new}")
```
