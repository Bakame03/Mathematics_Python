## Suites récurrentes

Dans ce TP, on étudie les suites récurrentes du type :
$$
  \begin{cases}
    u_0 \in \mathbb{R} \\ u_{n+1} = f(u_n)
  \end{cases}
$$
où $f$ est une fonction numérique supposée continue et dérivable sur un intervalle de $\mathbb{R}$.


**Exercice 1.** Écrire une fonction `U(f, n, a)` qui renvoie les $n+1$ premières valeurs de la suite $(u_n)$ sous la forme d'une liste.

```python
# Répondre à l'exercice 1 ici

def U(f, n, a) :
    u = a
    liste_u = [u]
    for k in range(1, n):
        u = f(u)
        liste_u.append(u)
    return liste_u
```

## Visualiser une suite récurrente

Le code des deux cellules suivantes permet de visualiser la trace de la suite $(u_n)$ comme nous l'avons vu en cours. Il convient de tracer sur le même graphique :
1. la courbe d'équation $y = f(x)$ ;
2. la bissectrice du repère (droite d'équation $y=x$) ;
3. la trace de la suite récurrente, c'est-à-dire, la ligne brisée joignant les points $\left( u_k; f(u_k) \right)$, $(u_{k+1}; u_{k+1})$ et $\left( u_{k+1} ; f(u_{k+1}) \right)$.

**Exercice 2.**  Compléter le code des cellules suivantes afin de visualiser la trace de la suite définie par :
$$
  \begin{cases}
    u_0 = a \\ u_{n+1} = \exp(-u_n) \quad \text{pour } n \geqslant 0
  \end{cases}
$$

On pendra $a=-1$ et $n=10$.

Quelles conjectures peut-on faire (sens de variation, limite) ?

```python
def liste_points(f, n, a):
    u = U(f, n, a)
    mespoints = [ (u[0],0) ]
    for k in range(n-1):
        mespoints.append( (u[k],u[k+1]) )
        mespoints.append( (u[k+1],u[k+1]) )
    return mespoints
```

```python
n = 10
a = -1

x = var('x')

# définir ici la fonction f
def f(x):
    return exp(-x)

mes_points = liste_points(f, n, a)

G = plot(f(x), -1, 3) # La fonction
G = G + plot(x, -1, 3, color='black') # La droite (y=x)
G = G + line(mes_points, color='red')
G.show()
```

**Exercice 3.**  De même, visualiser la trace de la suite définie par récurrence par $u_0 = \frac{5}{4}$ et $u_{n+1} = 2u_n-1$ avec $n = 10$.

```python
n = 10
a = 5/4
def f(x):
   return 2*x - 1
mes_points = liste_points(f, n, a)

G = plot(f(x), -1, 120)
G = G + plot(x, -1, 120, color='black')
G = G + line(mes_points, color='red')
G.show()
```

#### Conclusion. 

Les suites définies par récurrence peuvent être
- soit convergentes ;
- soit divergentes


Dans le cas où une suite est convergente vers un réel $\ell$, lorsque $n$ devient très grand, $u_n$ s'aproche de plus en plus vers $\ell$ mais aussi $u_{n+1}$ et la relation $u_{n+1} = f(u_n)$ devient
$$
  \ell = f(\ell)
$$
 
Cela signifie que :


**Propriété - Définition.**
Si une suite définie par la récurrence $u_{n+1} = f(u_n)$ admet une limite en $+\infty$, alors cette limite est **nécessairement** solution de l'équation
$$
  f(x) = x
$$
  Un nombre $x$ qui vérifie la relation $f(x) = x$ est appelé _point fixe_ de $f$.
